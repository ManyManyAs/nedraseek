# NedraSeek: запуск в Google Colab

1. На GitHub: **Code → Download ZIP**.
2. Здесь: панель «Файлы» слева, перетащить архив.
3. Выполнить ячейки по порядку.

```python
import os, glob, zipfile
if os.path.exists('../config.yaml'):          # ноутбук открыт из папки notebooks/ репозитория
    os.chdir('..')
elif not os.path.exists('config.yaml'):        # Colab: архив загружен в /content
    zips = sorted(glob.glob('/content/pipetwin-nedra*.zip'))
    if not zips:
        raise FileNotFoundError('Загрузите ZIP репозитория в панель «Файлы» слева и запустите ячейку снова')
    zipfile.ZipFile(zips[-1]).extractall('/content')
    os.chdir(sorted(glob.glob('/content/pipetwin-nedra*/'))[-1])
print(os.getcwd())
```
